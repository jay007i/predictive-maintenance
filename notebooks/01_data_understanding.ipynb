```python
unique_counts = df.nunique()

constant_features = unique_counts[unique_counts == 1].index.tolist()

print("Constant features:", constant_features)
print("Count:", len(constant_features))
```

```python
engine_51 = df[df["engine_id"] == 51]

plt.figure(figsize=(10,4))

plt.plot(
    engine_51["cycle"],
    engine_51["sensor_9"],
    marker=".",
    markersize=3
)

plt.xlabel("Operating Cycle")
plt.ylabel("Sensor 9")
plt.title("Sensor 9 Over Time - Engine 51")
plt.grid(True)
plt.show()
```

```python
engine_ids = [9,18,48,51]

plt.figure(figsize=(12,6))


for engine_id in engine_ids:
    engine_data = df[df["engine_id"] == engine_id]
    
    plt.plot(
        engine_data["cycle"],
        engine_data["sensor_9"],
        label =f"Engine {engine_id}"
    )
    
plt.xlabel("Operating Cycle")
plt.ylabel("Sensor 9")    
plt.title("Sensor 9 Trends Across Multiple Engines")
plt.legend()
plt.grid(True)
plt.show()
```
